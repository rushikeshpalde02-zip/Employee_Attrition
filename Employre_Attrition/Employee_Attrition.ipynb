```python
df6 = pd.get_dummies(df5).astype(int)
```

```python
# 1 #  Linear Regression
```

```python
#df6.head()
X = df6.drop(columns=['salary'])
y = df6['salary']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, y, train_size=0.20, random_state=42)
```

```python
#  Linear Regression
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# model = LinearRegression()
# model.fit(X_train[1:25000], y_train[1:25000])
#model.fit(X_train[1:20000], y_train[1:20000])
```

```python
%%time
# Fit a Linear Regression model to the train dataset

# Import LinearRegressor
from sklearn.linear_model import LinearRegression

# Instantiate the model
lModel = LinearRegression()


# Fit the model to the data
lModel.fit(X_train,Y_train)
```

```python
y_pred = lModel.predict(X_test)
```

```python
r2 = r2_score(Y_test,y_pred)
rmse = np.sqrt(mean_squared_error(Y_test,y_pred))

print(" R² Score:", r2)
print("RMSE:", rmse)
```

```python
# 2 # Decision Tree Regressor
```

```python
%%time
# Fit a Decision Tree Regressor model to the train dataset

# Import DecisionTreeRegressor
from sklearn.tree import DecisionTreeRegressor

# Instantiate the model
DTR = DecisionTreeRegressor()



# Instantiate the model
DTR.fit(X_train[1:50000],Y_train.iloc[1:50000])
```

```python
y_pred = DTR.predict(X_test)
```

```python
r2 = r2_score(Y_test,y_pred)
rmse = np.sqrt(mean_squared_error(Y_test,y_pred))

print(" R² Score:", r2)
print("RMSE:", rmse)
```
